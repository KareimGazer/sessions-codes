### Dirty Data

Dirty data = low quality data = issues regarding content

**Dirty Data**:	Data that has issues with data content including missing data, invalid data, inaccurate date or inconsistent data

1. **completeness**: many records are missing from our data, represented by Nan (not a number)
2. **Validity**: are the values in our records within acceptable ranges?
3. **accuracy**: inaccurate data is valid data but still wrong data (adheres to the schema (value ranges) but still incorrect) so the data maybe off by some margin due to problems in measurements like the tools used or scale or maybe wrong entry of the info
4. **Consistency**: inconsistent data is still valid and accurate but the representation of the records is still different meaning that we want a standard format for example the name of the state maybe written fully some times and other time is abbreviated which will make it hard to processes the data programmatically if the rules don't apply the same way to all data

Sources of Dirty Data
Dirty data = low quality data = content issues

There are lots of sources of dirty data. Basically, anytime humans are involved, there's going to be dirty data. There are lots of ways in which we touch data we work with.

- user entry errors.
- integrating data where different schemas (ranges) have been used for the same type of item.
- legacy data systems, where data wasn't coded when disc and memory constraints were much more restrictive than they are now. Over time systems evolve. Needs change, and data changes.
- Some of our data won't have the unique identifiers it should.
- data lost in the transformation from one format to another.
- programmer error.
- data might have been corrupted in transmission or storage by cosmic rays or other physical phenomena. So hey, one that's not our fault.

### Examples of Improving Quality

- **Correcting** when inaccurate, like correcting the body weight from 0.018 kg to 180 kg
- **Removing** when irrelevant, like removing the rows with incorrect info
- **Replacing** when missing, like filling in the missing valueS

## Data Cleaning


Cleaning: Acting on the assessments we made to improve quality and tidiness

Types of cleaning:
- **Manual** (not recommended unless the issues are single occurrences)
- **Programmatic**

### Manual Data Cleaning includes:

- Retyping incorrect data
- Copying and pasting columns and rows

However, manual cleaning is inefficient, error-prone, and demoralizing. So never clean manually.

### The programmatic data cleaning process:

- **Define**: convert our assessments into defined cleaning tasks. These definitions also serve as an instruction list so others (or yourself in the future) can look at your work and reproduce it.
- **Code**: convert those definitions to code and run that code.
- **Test**: test your dataset, visually or with code, to make sure your cleaning operations worked.

Programmatic Data Cleaning uses code to:

- Automate cleaning tasks
- Minimize repetition
- Save time

Data wrangling takes a tremendous amount of time for the data professional, so doing anything that saves time is great

Define:
- Select all records in the `StartDate` column that have `"As soon as possible"`, `"Immediately"`, etc. and replace the text in those cells with `"ASAP"`

```python
# code

asap_list = ['Immediately', 'As soon as possible', 'Upon hiring',
             'Immediate', 'Immediate employment', 'As soon as possible.', 'Immediate job opportunity',
             '"Immediate employment, after passing the interview."',
             'ASAP preferred', 'Employment contract signature date',
             'Immediate employment opportunity', 'Immidiately', 'ASA',
             'Asap', '"The position is open immediately but has a flexible start date depending on the candidates earliest availability."',
             'Immediately upon agreement', '20 November 2014 or ASAP',
             'immediately', 'Immediatelly',
             '"Immediately upon selection or no later than November 15, 2009."',
             'Immediate job opening', 'Immediate hiring', 'Upon selection',
             'As soon as practical', 'Immadiate', 'As soon as posible',
             'Immediately with 2 months probation period',
             '12 November 2012 or ASAP', 'Immediate employment after passing the interview',
             'Immediately/ upon agreement', '01 September 2014 or ASAP',
             'Immediately or as per agreement', 'as soon as possible',
             'As soon as Possible', 'in the nearest future', 'immediate',
             '01 April 2014 or ASAP', 'Immidiatly', 'Urgent',
             'Immediate or earliest possible', 'Immediate hire',
             'Earliest  possible', 'ASAP with 3 months probation period.',
             'Immediate employment opportunity.', 'Immediate employment.',
             'Immidietly', 'Imminent', 'September 2014 or ASAP', 'Imediately']

for phrase in asap_list:
    df_clean.StartDate.replace(phrase, 'ASAP', inplace=True)
```

**test**

```python
# test
df_clean.info()
```

df_clean.StartDate.value_counts()

**Assert statement**:	A debugging test that throws an error message if a logical statement evaluates to False

```python
# loop over each phrase in the asap_list and check that it is not contained in the values of the StartDate column
for phrase in asap_list:
    assert phrase not in df_clean.StartDate.values
```

## Analysis & Visualization

Insight: trying to find the percentage of positions with an urgent start date i.e. `'ASAP'`

```python
# Number of 'ASAP' start dates (numerator)
asap_counts = df_clean.StartDate.value_counts()['ASAP']
asap_counts
```

```python
# Number of non-empty start dates (denominator)
non_empty_counts = df_clean.StartDate.count()
non_empty_counts
```

```python
# Percentage of positions with an urgent start date i.e. 'ASAP'
asap_counts / non_empty_counts
```

```python
%matplotlib inline
df_clean.StartDate.value_counts().plot(kind="pie")
```
